# Part B: Dataset Understanding & Preparation

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Set visualization style
sns.set_style("whitegrid")

# Set random state
RANDOM_STATE = 42

print("Libraries imported successfully.")
```

### Dataset Loading

```python
df = pd.read_csv(r"C:\Users\tanaa\Downloads\Message_Intelligence_Dataset_5200 - Message_Intelligence_Dataset_5200.csv.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

### Missing Values and Duplicate Records

```python
print("Missing Values:")
print(df.isnull().sum())


print("\nDuplicate Records:")
print(df.duplicated().sum())
```

###  Data Preprocessing

```python
# Remove duplicate rows

df = df.drop_duplicates()

# Handle missing values

df = df.dropna()

print("Dataset shape after preprocessing:", df.shape)
```

```python

# Identify input features and target variable

X = df.drop(
    columns=["spam_label", "message_id", "message_text", "timestamp"]
)

y = df["spam_label"]

print("Input Features:")
print(X.columns.tolist())

print("\nTarget Variable:")
print(y.name)

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)
```

### Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

### Feature Scaling

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
```

# Part C: Baseline Model – K-Nearest Neighbors

```python
knn = KNeighborsClassifier(n_neighbors=5)

# Train the model

knn.fit(X_train_scaled, y_train)

# Make predictions

y_pred_knn = knn.predict(X_test_scaled)

print("KNN model trained successfully.")
```

```python
knn_accuracy = accuracy_score(y_test, y_pred_knn)

knn_precision = precision_score(
    y_test, y_pred_knn, zero_division=0
)

knn_recall = recall_score(
    y_test, y_pred_knn, zero_division=0
)

knn_f1 = f1_score(
    y_test, y_pred_knn, zero_division=0
)

print("KNN Evaluation Results")
print("----------------------")

print("Accuracy:", round(knn_accuracy, 4))
print("Precision:", round(knn_precision, 4))
print("Recall:", round(knn_recall, 4))
print("F1 Score:", round(knn_f1, 4))
```

```python
k_values = [3, 5, 7, 9, 11]

knn_results = []

for k in k_values:

    model = KNeighborsClassifier(n_neighbors=k)

    model.fit(X_train_scaled, y_train)

    predictions = model.predict(X_test_scaled)

    accuracy = accuracy_score(y_test, predictions)

    knn_results.append(accuracy)

    print(f"K = {k}, Accuracy = {accuracy:.4f}")
```

# Part D: Support Vector Machine (SVM)

```python
svm = SVC(kernel="rbf", random_state=42)

# Train the model

svm.fit(X_train_scaled, y_train)

# Make predictions

y_pred_svm = svm.predict(X_test_scaled)

print("SVM model trained successfully.")
```

```python
svm_accuracy = accuracy_score(y_test, y_pred_svm)

svm_precision = precision_score(
    y_test, y_pred_svm, zero_division=0
)

svm_recall = recall_score(
    y_test, y_pred_svm, zero_division=0
)

svm_f1 = f1_score(
    y_test, y_pred_svm, zero_division=0
)

print("SVM Evaluation Results")
print("----------------------")

print("Accuracy:", round(svm_accuracy, 4))
print("Precision:", round(svm_precision, 4))
print("Recall:", round(svm_recall, 4))
print("F1 Score:", round(svm_f1, 4))
```

```python

# Display classification report

print("SVM Classification Report")

print(
    classification_report(
        y_test,
        y_pred_svm,
        target_names=["Legitimate", "Spam"],
        zero_division=0
    )
)
```

# Part E: Naive Bayes Classifier

```python
from sklearn.naive_bayes import GaussianNB

nb = GaussianNB()

# Train the model

nb.fit(X_train_scaled, y_train)

# Make predictions

y_pred_nb = nb.predict(X_test_scaled)

print("Naive Bayes model trained successfully.")
```

```python
nb_accuracy = accuracy_score(y_test, y_pred_nb)

nb_precision = precision_score(
    y_test, y_pred_nb, zero_division=0
)

nb_recall = recall_score(
    y_test, y_pred_nb, zero_division=0
)

nb_f1 = f1_score(
    y_test, y_pred_nb, zero_division=0
)

print("Naive Bayes Evaluation Results")
print("-----------------------------")

print("Accuracy:", round(nb_accuracy, 4))
print("Precision:", round(nb_precision, 4))
print("Recall:", round(nb_recall, 4))
print("F1 Score:", round(nb_f1, 4))
```

```python
print("Naive Bayes Classification Report")

print(
    classification_report(
        y_test,
        y_pred_nb,
        target_names=["Legitimate", "Spam"],
        zero_division=0
    )
)
```

# Part F: Model Comparison

```python
comparison_df = pd.DataFrame({
    "Model": ["KNN", "SVM", "Naive Bayes"],

    "Accuracy": [
        knn_accuracy,
        svm_accuracy,
        nb_accuracy
    ],

    "Precision": [
        knn_precision,
        svm_precision,
        nb_precision
    ],

    "Recall": [
        knn_recall,
        svm_recall,
        nb_recall
    ],

    "F1 Score": [
        knn_f1,
        svm_f1,
        nb_f1
    ]
})

comparison_df
```
